```python
from typing import List
from network_module import Network, AbstractNode, Message
import networkx as nx
```

```python
class NeighbourBroadcastNode(AbstractNode):
    def __init__(self, node_id: int, neighbours: List[int]):
        super().__init__(node_id, neighbours)
        self.state["neighbour_ids"] = []

    def handle_message(self, message: Message) -> None:
        self.state["neighbour_ids"].append(message.content)

    def compute(self):
        for neighbour in self._neighbours:
            self.send_message(neighbour, self.node_id)
```

```python
G = nx.Graph()
G.add_nodes_from([1, 2, 3, 4])
G.add_edges_from([(1, 2), (1, 3), (2, 3), (2, 4), (3, 4)])
```

```python
network = Network(G, NeighbourBroadcastNode)
network.run(2)
print(network)
```
