```python
from network_module import WeightedNetwork, AbstractNode
import networkx as nx
```

```python
class NeighbourBroadcastMaxWeightEdge(AbstractNode):
    def __init__(self, node_id, neighbours):
        super().__init__(node_id, neighbours)
        self.state["neighbour_edge_weights"] = []
        self.state["max_edge"] = 0

    def handle_message(self, message):
        self.state["neighbour_edge_weights"].append(message.content)

    def compute(self):
        for neighbour in self._neighbours:
            self.send_message(neighbour, list(self._neighbours.values()))

        if self.state["neighbour_edge_weights"]:
            max_value = max(map(max, self.state["neighbour_edge_weights"]))
            self.state["max_edge"] = max_value
```

```python
# Create an empty graph
G = nx.Graph()

# Add nodes
G.add_nodes_from([1, 2, 3, 4])

# Add edges with weights
G.add_edge(1, 2, weight=1.5)
G.add_edge(1, 3, weight=2.0)
G.add_edge(2, 3, weight=2.5)
G.add_edge(2, 4, weight=1.0)
G.add_edge(3, 4, weight=3.0)
```

```python
network = WeightedNetwork(G, NeighbourBroadcastMaxWeightEdge)
network.run(2)
print(network)
```
